```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

from tensorflow.keras.models import load_model
from tensorflow.keras.callbacks import EarlyStopping,ModelCheckpoint
```

```python
x_train=np.load("../processed/x_train_sequence.np.npy")
```

```python
x_train.shape
```

```python
x_train_split,x_val=train_test_split(x_train,test_size=0.2,random_state=42,shuffle=True)
```

```python
model=load_model("../models/lstm_autoencoder.keras")
```

```python
model.summary()
```

```python
early_stopping=EarlyStopping(monitor="val_loss",patience=5,restore_best_weights=True,verbose=1)
```

```python
check_point=ModelCheckpoint("../models/lstm_autoencoder.keras",monitor="val_loss",save_best_only=True,verbose=1)
```

```python
history=model.fit(
    x_train_split,
    x_train_split,
    epochs=100,
    batch_size=128,
    shuffle=True,
    validation_data=(x_val,x_val),
    callbacks=[early_stopping,check_point]
)
```

```python
model.save("../models/lstm_autoencoder_final.keras")
```
