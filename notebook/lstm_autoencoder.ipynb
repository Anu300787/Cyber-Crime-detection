```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,RepeatVector,TimeDistributed,Dense
```

```python
x_train=np.load("../processed/X_train_sequence.npy")
```

```python
x_train.shape
```

```python
timesteps = x_train.shape[1]
features = x_train.shape[2]
```

```python
model=Sequential()
```

```python
model.add(
    LSTM(units=64,activation="relu",input_shape=(timesteps,features),return_sequences=True)
    )
```

```python
model.add(
    LSTM(units=32,activation="relu",return_sequences=False)
)
```

```python
model.add(RepeatVector(timesteps))
```

```python
model.add(LSTM(
    units=32,
    activation="relu",
    return_sequences=True
)
)
```

```python
model.add(LSTM(
    units=64,
    activation="relu",
    return_sequences=True
)
)
```

```python
model.add(TimeDistributed(Dense(features)))
```

```python
from tensorflow.keras.optimizers import Adam
optimizer=Adam(learning_rate=0.001)
model.compile(optimizer="Adam",loss="MSE")
```

```python
model.summary()
```

```python
model.save("../models/lstm_autoencoder.keras")
```
