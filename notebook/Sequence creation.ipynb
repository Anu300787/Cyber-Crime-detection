```python
import pandas as pd
import numpy as np 
```

```python
x_train_scaled=pd.read_csv("../data/scaled_benign.csv")
scaled_df=pd.read_csv("../data/scaled_network_flow.csv")
```

```python
X_all_scaled=scaled_df.drop("Label",axis=1)
y=scaled_df["Label"]
```

```python
x_train_scaled.shape
```

```python
sequence_len=30
```

```python
def create_sequences(X, sequence_length, labels=None):

    X_sequences = []
    y_sequences = []

    for i in range(len(X) - sequence_length + 1):

        X_sequences.append(
            X[i:i + sequence_length]
        )

        if labels is not None:

            y_sequences.append(
                labels.iloc[i + sequence_length - 1]
            )

    X_sequences = np.array(
        X_sequences,
        dtype=np.float32
    )

    if labels is not None:

        return (
            X_sequences,
            np.array(y_sequences)
        )

    return X_sequences
```

```python
x_train_seq=create_sequences(x_train_scaled,sequence_len)
```

```python
x_train_seq.shape
```

```python
X_test_sequence, y_test = create_sequences(
    X_all_scaled,
    sequence_length=30,
    labels=y
)
```

```python
print("sample shape",x_train_seq[0].shape)
print("timestamp",x_train_seq[1].shape)
print("feature shape",x_train_seq[2].shape)
```

```python
np.save(
    "../processed/X_train_sequence.npy",
    x_train_seq.astype(np.float32)
)

np.save(
    "../processed/X_test_sequence.npy",
    X_test_sequence.astype(np.float32)
)

np.save(
    "../processed/y_test.npy",
    y_test
)
```
