```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv("../data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv")
```

```python
df.isnull().sum()
```

```python
np.isinf(df.select_dtypes(include=[np.number])).sum()
```

```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```
